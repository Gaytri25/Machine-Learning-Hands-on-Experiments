```python

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd

# Bread and butter shopping transactions
transactions = [
    ['Bread', 'Butter', 'Milk'],
    ['Bread', 'Butter'],
    ['Bread', 'Milk'],
    ['Butter', 'Milk'],
    ['Bread', 'Butter', 'Jam'],
    ['Bread', 'Butter', 'Milk'],
    ['Bread', 'Jam'],
    ['Bread', 'Butter'],
    ['Bread', 'Milk', 'Jam'],
        ['Bread', 'Milk'],
        ['Bread', 'Bread'],

]

# Convert transactions into a binary dataset
te = TransactionEncoder()
encoded = te.fit(transactions).transform(transactions)
df = pd.DataFrame(encoded, columns=te.columns_)

print("TRANSACTION DATASET")
print(df.astype(int))


frequent = apriori(df, min_support=0.25, use_colnames=True)
print("\nFREQUENT ITEMSETS")
print(frequent)

#  association rules
rules = association_rules(
    frequent, metric="confidence", min_threshold=0.6
)
print("\nASSOCIATION RULES")
print(rules[['antecedents', 'consequents',
             'support', 'confidence', 'lift']])

# Identify outliers using the IQR method
data = pd.DataFrame({
    'Daily_Sales': [100, 110, 105, 115, 120, 108, 112, 500,800,200,12,39,82,92,0,98]
})

Q1 = data['Daily_Sales'].quantile(0.25)
Q3 = data['Daily_Sales'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[
    (data['Daily_Sales'] < lower) |
    (data['Daily_Sales'] > upper)
]

print("\nOUTLIERS")
print(outliers)

print("\nOUTLIERS in data set ")



```

```python
import os
import pandas as pd

# Find the CSV file inside the downloaded kagglehub directory
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    print(f"Loading file from: {csv_path}")

    # Load dataset
    salary_df = pd.read_csv(csv_path)
    display(salary_df.head())

    # Calculate IQR and detect outliers in the 'Salary' column
    # Checking column name variation (e.g. 'Salary' or 'salary')
    col_name = 'Salary' if 'Salary' in salary_df.columns else salary_df.columns[1] # fallback to the second column

    data_series = salary_df[col_name]
    Q1 = data_series.quantile(0.25)
    Q3 = data_series.quantile(0.75)
    IQR = Q3 - Q1

    outliers_salary = salary_df[(data_series < Q1 - 1.5 * IQR) | (data_series > Q3 + 1.5 * IQR)]
    print(f"\nDetected Outliers in {col_name} Column:")
    display(outliers_salary)
else:
    print("No CSV file found in the downloaded dataset directory.")
```
